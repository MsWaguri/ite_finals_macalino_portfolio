```python

import pandas as pd
import matplotlib.pyplot as plt

# Read data from CSV file
df = pd.read_csv("sales_data.csv")

# Display the data
print(df)

# Create a line chart
plt.plot(df["Month"], df["Sales"],
         marker="o", label="Sales", color="blue")

plt.plot(df["Month"], df["Expenses"],
         marker="s", label="Expenses", color="red")

# Add chart information
plt.title("Monthly Sales and Expenses")
plt.xlabel("Month")
plt.ylabel("Amount ($)")
plt.legend()
plt.grid(True)

# Display chart
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# URL of online CSV dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/flights.csv"

# Read data directly from the website
df = pd.read_csv(url)

# Convert year and month into dates
df["Date"] = pd.to_datetime(
    df["year"].astype(str) + "-" + df["month"],
    format="%Y-%B"
)

# Display first 10 records
print(df.head(10))

# Create line chart
plt.figure(figsize=(12, 6))

plt.plot(
    df["Date"],
    df["passengers"],
    color="blue",
    marker="o",
    markersize=3,
    linewidth=1.5
)

# Chart title and labels
plt.title("Airline Passenger Statistics (1949-1960)")
plt.xlabel("Year")
plt.ylabel("Number of Passengers (Thousands)")

# Add grid
plt.grid(True)

# Improve chart layout
plt.tight_layout()

# Display chart
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Read Excel file
df = pd.read_excel("company_data.xlsx")

# Display records
print(df.head())

# Create bar chart
plt.figure(figsize=(10, 5))

plt.bar(df["Department"], df["Employees"],
        color="orange")

plt.title("Employees per Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
