```python
students = [
    ("Ana", (85, 90, 88)),
    ("Ben", (78, 82, 88)),
    ("Carlo", (92, 95, 91)),
    ("Eric", (70, 75, 72)),
    ("Faith", (95, 93, 97))
]

print("STUDENT GRADE REPORT")

averages = []


for name, grades in students:
    average = sum(grades) / len(grades)
    averages.append((name, average))
    print(f"{name}: Average = {average:.2f}")


top_student = max(averages, key=lambda student: student[1])

print(f"\nTop Student: {top_student[0]}")


qualified = []

for name, average in averages:
    if average >= 85:
        qualified.append(name)

print("Qualified Students:", qualified)


averages.sort(key=lambda student: student[1], reverse=True)

print("\nSorted Results:")
print(averages)
```

```python
numbers = [
    12, 5, 8, 12, 3, 5, 9, 8,
    15, 3, 20, 12, 9, 25, 15
]

unique_numbers = []

for num in numbers:
    if num not in unique_numbers:
        unique_numbers.append(num)


counts = {}

for num in numbers:
    counts[num] = counts.get(num, 0) + 1


max_count = max(counts.values())

most_frequent = []

for num, count in counts.items():
    if count == max_count:
        most_frequent.append(num)


unique_tuple = tuple(unique_numbers)


print("Original List:", numbers)
print("Unique List:", unique_numbers)

print("\nFrequency Report:")
for num, count in counts.items():
    print(f"Number {num}: appeared {count} time(s)")

print("\nMost Frequently Occurring Number(s):",
      most_frequent, f"(Appeared {max_count} times)")

print("Unique Tuple:", unique_tuple)
```

```python
inventory = [
    ("KB101", "Keyboard", 25, 450.00),
    ("MS102", "Mouse", 40, 250.00),
    ("HD103", "Hard Drive", 12, 2800.00),
    ("MN104", "Monitor", 8, 6500.00),
    ("USB105", "USB Drive", 50, 350.00)
]

summary = [(name, qty, qty * price) for code, name, qty, price in inventory]

print("Total Value per Product:")

for name, qty, value in summary:
    print(f"{name}: {value:.2f}")


total = sum(value for name, qty, value in summary)

print(f"\nTotal Inventory Value: {total:.2f}")

low_stock = [name for code, name, qty, price in inventory if qty < 15]

print("Products with fewer than 15 units:", low_stock)

highest = max(summary, key=lambda x: x[2])

print("Product with highest total inventory value:", highest[0])

sorted_inventory = sorted(inventory, key=lambda x: x[3], reverse=True)

print("\nInventory sorted by unit price (descending):")

print(sorted_inventory)

print("\nNew list of tuples (Product Name, Quantity, Total Value):")

print(summary)

search_code = input("\nEnter product code to search: ").upper()

found = False

for code, name, qty, price in inventory:
    if code.upper() == search_code:
        print(f"Product Found: Code={code}, Name={name}, Quantity={qty}, Unit Price={price:.2f}")
        found = True
        break

if not found:
    print("Product code does not exist.")
```

```python
contestants = [
    ("C001", "Alex", (85, 90, 88)),
    ("C002", "Bea", (95, 80, 92)),
    ("C003", "Chris", (90, 90, 90)),
    ("C004", "Dana", (95, 85, 87)),
    ("C005", "Evan", (88, 92, 91)),
    ("C006", "Gia", (95, 85, 87))
]

leaderboard = []
qualified = []

for c_id, name, scores in contestants:
    total = sum(scores)
    avg = total / len(scores)
    highest = max(scores)
    leaderboard.append((c_id, name, total, avg, highest))

    if avg >= 90:
        qualified.append(name)

leaderboard.sort(key=lambda x: (x[2], x[4]), reverse=True)

final_leaderboard = [
    (c_id, name, total, round(avg, 2))
    for c_id, name, total, avg, highest in leaderboard
]

print("LEADERBOARD:")
for rank, (c_id, name, total, avg) in enumerate(final_leaderboard, 1):
    print(f"Rank {rank}: ID: {c_id} | Name: {name} | Total Score: {total} | Average Score: {avg:.2f}")

print("\nTOP 3 CONTESTANTS:")
for rank, (c_id, name, total, avg) in enumerate(final_leaderboard[:3], 1):
    print(f"Rank {rank}: {name} (Total: {total}, Average: {avg:.2f})")

print("\nContestants with average score of at least 90:")
print(qualified)

print("\nFinal Leaderboard (List of Tuples):")
print(final_leaderboard)
```

```python
data = [
    14, 22, 18, 14, 35, 27, 22, 40,
    18, 14, 31, 27, 45, 22, 35, 50,
    18, 40, 27, 31
]

def sort_data(data):
    return sorted(data)

def remove_duplicates(data):
    return list(dict.fromkeys(data))

def frequency_table(data):
    counts = {num: data.count(num) for num in remove_duplicates(data)}
    print("\nFrequency Table:")
    for num, count in counts.items():
        print(f"Value {num}: {count} time(s)")

    highest = max(counts.values())
    most_frequent = [num for num, count in counts.items() if count == highest]
    print(f"Highest Frequency Value(s) (appeared {highest} times):", most_frequent)

def display_min_max(data):
    print("\nMinimum Value:", min(data))
    print("Maximum Value:", max(data))

def display_mean_median(data):
    sorted_data = sort_data(data)
    n = len(data)
    mean = sum(data) / n

    if n % 2 == 1:
        median = sorted_data[n // 2]
    else:
        median = (sorted_data[n // 2 - 1] + sorted_data[n // 2]) / 2

    print(f"\nMean: {mean:.2f}")
    print("Median:", median)

def display_second_largest(data):
    unique = sort_data(remove_duplicates(data))
    if len(unique) >= 2:
        print("\nSecond-Largest Distinct Value:", unique[-2])
    else:
        print("\nNot enough distinct values to find second-largest.")

def search_value(data):
    user_input = input("\nEnter a number to search: ")

    if user_input.lstrip('-').isdigit():
        target = int(user_input)
        count = data.count(target)

        if count:
            print(f"Value {target} was found in the dataset {count} time(s).")
        else:
            print(f"Value {target} is not in the dataset.")
    else:
        print("Invalid input. Please enter a valid numeric value.")

while True:
    print("\n----- DATA ANALYZER -----")
    print("1. Display Original Data")
    print("2. Display Sorted Data")
    print("3. Remove Duplicates")
    print("4. Display Frequency Table")
    print("5. Display Minimum and Maximum")
    print("6. Display Mean and Median")
    print("7. Display Second-Largest Distinct Value")
    print("8. Convert Data into a Tuple")
    print("9. Search for a Value")
    print("10. Exit")

    choice = input("\nEnter your choice (1-10): ")

    if choice == '1':
        print("\nOriginal Data:")
        print(data)
    elif choice == '2':
        print("\nSorted Data:")
        print(sort_data(data))
    elif choice == '3':
        print("\nData without Duplicates:")
        print(remove_duplicates(data))
    elif choice == '4':
        frequency_table(data)
    elif choice == '5':
        display_min_max(data)
    elif choice == '6':
        display_mean_median(data)
    elif choice == '7':
        display_second_largest(data)
    elif choice == '8':
        print("\nConverted Data into Tuple:")
        print(tuple(data))
    elif choice == '9':
        search_value(data)
    elif choice == '10':
        print("\nExiting program...")
        break
    else:
        print("\nInvalid choice. Please enter a number from 1 to 10.")
```
