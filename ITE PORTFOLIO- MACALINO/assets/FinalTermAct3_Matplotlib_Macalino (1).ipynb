```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [1, 4, 2, 3]

plt.plot(x, y)
plt.title("Simple Line Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = [0, 1, 2, 3]

y1 = [3, 8, 1, 10]
y2 = [6, 2, 7, 11]

plt.plot(x, y1, label="Line 1")
plt.plot(x, y2, label="Line 2")

plt.title("Multiple Line Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6]
y = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x, y, scale = np.random.rand(3, 100)

fig, ax = plt.subplots()

ax.scatter(x=x, y=y, c=scale, s=np.abs(scale)*500)

ax.set_title("Random Data Scatter Plot")
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = ["A", "B", "C", "D"]
y = [3, 8, 1, 10]

plt.bar(x, y)
plt.title("Bar Chart")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.normal(170, 10, 250)

plt.hist(x)
plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

y = [35, 25, 25, 15]

labels = ["A", "B", "C", "D"]

plt.pie(y, labels=labels, autopct="%1.1f%%")
plt.title("Pie Chart")
plt.show()
```
