```python

# ==========================================
# BIRD-SHAPED PYTHON WORD CLOUD USING CSV
# ==========================================

# 1. INSTALL LIBRARIES
!pip install -q wordcloud matplotlib numpy pandas

# 2. IMPORT LIBRARIES
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from wordcloud import WordCloud, STOPWORDS
from google.colab import files

# 3. UPLOAD CSV MASK
uploaded = files.upload()

# 4. READ CSV FILE
filename = next(iter(uploaded))

mask = pd.read_csv(
    filename,
    header=None
).values.astype(np.uint8)

# 5. CHECK MASK
print("CSV successfully loaded!")
print("Mask dimensions:", mask.shape)

# Display mask preview
plt.figure(figsize=(8, 5))
plt.imshow(mask, cmap="gray")
plt.title("Bird Mask Preview")
plt.axis("off")
plt.show()

# 6. PYTHON WORD CLOUD TEXT
text = """
Python Python Python Python Python Python
programming programming programming programming
widely widely widely
used used used used used
learn learn learn learning
machine machine machine
web web web
language language language language
versatile versatile versatile
powerful powerful powerful powerful powerful
data data data data
easy easy easy easy
learning learning learning
science science science
development development development
coding coding coding
computer computer computer
software software software
technology technology technology
artificial intelligence
programming language
computer science
data analysis
machine learning
web development
"""

# 7. SET STOPWORDS
stopwords = set(STOPWORDS)

# 8. GENERATE WORD CLOUD
wordcloud = WordCloud(
    background_color="white",
    mask=mask,
    max_words=250,
    max_font_size=90,
    min_font_size=3,
    stopwords=stopwords,
    contour_width=2,
    contour_color="black",
    colormap="viridis",
    relative_scaling=0.5,
    collocations=False,
    random_state=42,
    scale=3
).generate(text)

# 9. DISPLAY FINAL WORD CLOUD
plt.figure(figsize=(12, 9), dpi=150)

plt.imshow(
    wordcloud,
    interpolation="bilinear"
)

plt.axis("off")
plt.title("Python Word Cloud", fontsize=18)
plt.tight_layout(pad=0)
plt.show()

# 10. SAVE IMAGE
wordcloud.to_file("Bird_Python_WordCloud.png")

# 11. DOWNLOAD FINAL IMAGE
files.download("Bird_Python_WordCloud.png")
```
